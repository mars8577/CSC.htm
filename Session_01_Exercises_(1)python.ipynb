# Section 3: Arithmetic Operations

## Exercise 3.1: Rectangle Calculator

Calculate properties of a rectangle.

```python
# Exercise 3.1: Rectangle Calculator

# Given dimensions
length = 15.5  # meters
width = 8.0    # meters

# Calculate area (length × width)
area = (length * width)

# Calculate perimeter (2 × length + 2 × width)
perimeter =( length **2 + width **2)

# Calculate diagonal using Pythagorean theorem: √(length² + width²)
# Hint: square root can be calculated as ** 0.5
diagonal = (length + width) ** 0.5

print(f"Rectangle: {length}m × {width}m")
print(f"Area: {area} square meters")
print(f"Perimeter: {perimeter} meters")
print(f"Diagonal: {diagonal:.2f} meters")
```

## Exercise 3.2: Temperature Converter

Convert between Fahrenheit and Celsius.

```python
# Exercise 3.2: Temperature Converter

# Formula: Celsius = (Fahrenheit - 32) × 5/9
# Formula: Fahrenheit = Celsius × 9/5 + 32

# Part 1: Convert these Fahrenheit values to Celsius
temp_f1 = 98.6   # Normal body temperature
temp_f2 = 212    # Boiling point of water
temp_f3 = 32     # Freezing point of water

temp_c1 = (temp_f1 - 32) * 5/9
temp_c2 = (temp_f2 - 32) * 5/9
temp_c3 = (temp_f3 - 32) * 5/9

print(f"{temp_f1}°F = {temp_c1:.1f}°C")
print(f"{temp_f2}°F = {temp_c2:.1f}°C")
print(f"{temp_f3}°F = {temp_c3:.1f}°C")

print()  # Empty line for readability

# Part 2: Convert these Celsius values to Fahrenheit
temp_c4 = 0      # Freezing
temp_c5 = 25     # Room temperature
temp_c6 = 100    # Boiling

temp_f4 = (temp_c4 * 9/5) + 32
temp_f5 = (temp_c5 * 9/5) + 32
temp_f6 = (temp_c6 * 9/5) + 32

print(f"{temp_c4}°C = {temp_f4:.1f}°F")
print(f"{temp_c5}°C = {temp_f5:.1f}°F")
print(f"{temp_c6}°C = {temp_f6:.1f}°F")
```

## Exercise 3.3: Tip Calculator

Calculate tips and split bills.

```python
# Exercise 3.3: Tip Calculator

# Scenario: Dinner with friends
meal_cost = 84.50
tax_rate = 0.0875  # 8.75%
tip_rate = 0.20    # 20%
num_people = 4

# Step 1: Calculate tax amount
tax_amount = meal_cost * tax_rate

# Step 2: Calculate subtotal (meal + tax)
subtotal = meal_cost + tax_amount

# Step 3: Calculate tip (based on meal cost before tax)
tip_amount = meal_cost + tip_rate

# Step 4: Calculate total bill
total_bill = subtotal+ tip_rate

# Step 5: Calculate amount per person
per_person = total_bill / num_people

print("=== Bill Summary ===")
print(f"Meal cost:    ${meal_cost:.2f}")
print(f"Tax (8.75%):  ${tax_amount:.2f}")
print(f"Tip (20%):    ${tip_amount:.2f}")
print(f"─────────────────────")
print(f"Total:        ${total_bill:.2f}")
print(f"Per person:   ${per_person:.2f}")
```

## Exercise 3.4: Compound Interest Calculator

Calculate investment growth with compound interest.

```python
# Exercise 3.4: Compound Interest

# Formula: A = P × (1 + r)^n
# Where:
#   A = final amount
#   P = principal (initial investment)
#   r = annual interest rate (as decimal)
#   n = number of years

# Scenario: You invest $5,000 at 7% annual interest
principal = 5000
annual_rate = 0.07

# Calculate the balance after 1, 5, 10, and 20 years
balance_1yr = principal * (1 * annual_rate) **1
balance_5yr = principal * (1 * annual_rate) **5
balance_10yr = principal * (1 * annual_rate) **10
balance_20yr = principal * (1 * annual_rate) **20

print(f"Initial investment: ${principal:,.2f}")
print(f"Annual rate: {annual_rate*100}%")
print()
print(f"After 1 year:   ${balance_1yr:,.2f}")
print(f"After 5 years:  ${balance_5yr:,.2f}")
print(f"After 10 years: ${balance_10yr:,.2f}")
print(f"After 20 years: ${balance_20yr:,.2f}")

# Bonus: Calculate total interest earned after 20 years
interest_earned = balance_20yr - principal
print(f"\nTotal interest earned: ${interest_earned:,.2f}")
```

## Exercise 4.2: Leap Year Checker

Determine if a year is a leap year using comparisons.

**Leap Year Rules:**
- Divisible by 4 → leap year
- EXCEPT divisible by 100 → NOT a leap year
- EXCEPT divisible by 400 → IS a leap year

**Hint:** Use the modulo operator `%` to check divisibility. If `year % 4 == 0`, the year is divisible by 4.

```python
# Exercise 4.2: Leap Year Checker

# Test with different years
year = 2024

# Check divisibility
divisible_by_4 = year % 4 == 0    # year % 4 == 0
divisible_by_100 = year % 100 == 0  # year % 100 == 0
divisible_by_400 = year % 400 == 0  # year % 400 == 0

# Leap year logic:
# Is leap year if: (divisible by 4 AND NOT divisible by 100) OR (divisible by 400)
is_leap_year = ("divisible by 4 and not divisible by 100") or ("divisible by 400")

print(f"Year: {year}")
print(f"Divisible by 4: {divisible_by_4}")
print(f"Divisible by 100: {divisible_by_100}")
print(f"Divisible by 400: {divisible_by_400}")
print(f"Is leap year: {is_leap_year}")

# Test your logic with these years:
# 2024 → True (divisible by 4)
# 1900 → False (divisible by 100, not 400)
# 2000 → True (divisible by 400)
# 2023 → False (not divisible by 4)
```

# Section 6: Real-World Data Scenarios

## Exercise 6.1: Sales Performance Analysis

```python
# Exercise 6.1: Sales Performance

# Monthly sales data for a salesperson
january_sales = 45000
february_sales = 52000
march_sales = 38000
monthly_quota = 40000

# Calculate:
# 1. Total Q1 sales
total_q1_sales = january_sales + february_sales + march_sales

# 2. Average monthly sales
average_monthly = total_q1_sales / 3

# 3. Highest sales month value
# Hint: Use max() function - max(a, b, c)
highest_month = max(january_sales, february_sales, march_sales)

# 4. Did they meet quota each month?
jan_met_quota = january_sales >= monthly_quota
feb_met_quota = february_sales >= monthly_quota
mar_met_quota = march_sales >= monthly_quota

# 5. How many months met quota?
# Hint: In Python, True = 1 and False = 0 when used in math
months_at_quota = jan_met_quota + feb_met_quota + mar_met_quota

# 6. Percentage over/under Q1 quota (quota × 3 for quarter)
quarterly_quota = monthly_quota * 3
quota_percentage = (total_q1_sales / quarterly_quota) * 100  # (total_q1_sales / quarterly_quota) * 100

print("=== Q1 Sales Report ===")
print(f"Total Q1 Sales: ${total_q1_sales:,}")
print(f"Average Monthly: ${average_monthly:,.2f}")
print(f"Best Month: ${highest_month:,}")
print(f"Months at Quota: {months_at_quota}/3")
print(f"Q1 Quota Achievement: {quota_percentage:.1f}%")
```

## Exercise 6.2: E-commerce Metrics

```python
# Exercise 6.2: E-commerce Dashboard Metrics

# Website data for last month
website_visitors = 125000
page_views = 450000
total_orders = 3750
total_revenue = 187500.00
returning_customers = 1500
new_customers = 2250
cart_abandonment_count = 8500  # People who added items but didn't buy

# Calculate these KPIs:

# 1. Conversion Rate: (orders / visitors) × 100
conversion_rate = (total_orders / website_visitors) * 100

# 2. Average Order Value (AOV): revenue / orders
aov = total_revenue / total_orders

# 3. Pages per Visit: page_views / visitors
pages_per_visit = page_views / website_visitors

# 4. Returning Customer Rate: (returning / total orders) × 100
returning_rate = (returning_customers / total_orders) * 100

# 5. Cart Abandonment Rate: (abandoned / (abandoned + orders)) × 100
cart_abandonment_rate = (cart_abandonment_count / (cart_abandonment_count + total_orders ))

# 6. Revenue per Visitor
revenue_per_visitor = total_revenue / website_visitors

# Display as a dashboard
print("╔═══════════════════════════════════════╗")
print("║       E-COMMERCE DASHBOARD            ║")
print("╠═══════════════════════════════════════╣")
print(f"║ Visitors:              {website_visitors:>12,} ║")
print(f"║ Orders:                {total_orders:>12,} ║")
print(f"║ Revenue:               ${total_revenue:>11,.2f} ║")
print("╠═══════════════════════════════════════╣")
print(f"║ Conversion Rate:       {conversion_rate:>11.2f}% ║")
print(f"║ Avg Order Value:       ${aov:>11.2f} ║")
print(f"║ Pages per Visit:       {pages_per_visit:>12.1f} ║")
print(f"║ Returning Customer:    {returning_rate:>11.1f}% ║")
print(f"║ Cart Abandonment:      {cart_abandonment_rate:>11.1f}% ║")
print(f"║ Revenue per Visitor:   ${revenue_per_visitor:>11.2f} ║")
print("╚═══════════════════════════════════════╝")
```

## Challenge 7.1: Movie Theater Revenue

```python
# Challenge 7.1: Movie Theater Revenue Calculator

# Ticket prices
adult_price = 15.00
child_price = 10.00
senior_price = 12.00

# Tickets sold for today
adult_tickets = 245
child_tickets = 87
senior_tickets = 43

# Concession average per customer
concession_avg = 8.50
concession_purchase_rate = 0.65  # 65% of customers buy concessions

# Calculate:
# 1. Total tickets sold
total_tickets = adult_price + child_price + senior_price

# 2. Revenue from each ticket type
adult_revenue = adult_price * adult_tickets
child_revenue = child_price * child_tickets
senior_revenue = senior_price * senior_tickets

# 3. Total ticket revenue
ticket_revenue = adult_revenue + child_revenue + senior_revenue

# 4. Estimated concession revenue
# (total customers × purchase rate × average spend)
concession_revenue = (total_tickets * concession_purchase_rate) * concession_avg

# 5. Total daily revenue
total_revenue = ticket_revenue + concession_revenue

# 6. Average revenue per customer
avg_per_customer = ticket_revenue / total_tickets

print("=== Daily Revenue Report ===")
print(f"\nTicket Sales:")
print(f"  Adult ({adult_tickets}):   ${adult_revenue:,.2f}")
print(f"  Child ({child_tickets}):    ${child_revenue:,.2f}")
print(f"  Senior ({senior_tickets}):   ${senior_revenue:,.2f}")
print(f"  ─────────────────────")
print(f"  Subtotal:       ${ticket_revenue:,.2f}")
print(f"\nConcessions:      ${concession_revenue:,.2f}")
print(f"\n══════════════════════════")
print(f"TOTAL REVENUE:    ${total_revenue:,.2f}")
print(f"Avg per Customer: ${avg_per_customer:.2f}")
```

## Challenge 7.2: Inventory Analysis

```python
# Challenge 7.2: Inventory Health Check

# Inventory for a product
product_name = "Wireless Earbuds"
current_stock = 145
reorder_point = 50      # Order more when stock falls to this level
max_capacity = 500      # Maximum warehouse can hold
daily_sales_avg = 12    # Average units sold per day
lead_time_days = 7      # Days to receive new shipment
unit_cost = 25.00       # Cost per unit
unit_price = 49.99      # Selling price

# Calculate:
# 1. Days of inventory remaining (current_stock / daily_sales_avg)
days_remaining = (current_stock / daily_sales_avg)

# 2. Units needed during lead time (daily_sales × lead_time)
units_during_lead_time = (daily_sales_avg * lead_time_days)

# 3. Should we reorder? (current_stock <= reorder_point)
should_reorder = (current_stock <= reorder_point)

# 4. Stock level percentage (current_stock / max_capacity × 100)
stock_level_pct = (current_stock / max_capacity) * 100

# 5. Current inventory value (cost basis)
inventory_value = current_stock * unit_cost

# 6. Potential revenue if all sold
potential_revenue = current_stock * unit_price

# 7. Potential profit
potential_profit = potential_revenue - inventory_value

# 8. Is stock critically low? (less than lead_time worth of sales)
is_critical = current_stock <= units_during_lead_time

print(f"=== Inventory Report: {product_name} ===")
print(f"\nCurrent Stock: {current_stock} units ({stock_level_pct:.1f}% capacity)")
print(f"Days of Stock: {days_remaining:.1f} days")
print(f"Reorder Point: {reorder_point} units")
print(f"\nInventory Value: ${inventory_value:,.2f}")
print(f"Potential Revenue: ${potential_revenue:,.2f}")
print(f"Potential Profit: ${potential_profit:,.2f}")
print(f"\n⚠️  Reorder Needed: {should_reorder}")
print(f"🚨 Critical Level: {is_critical}")
```
