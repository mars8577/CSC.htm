LIST COMPREHENSIONS
List comprehension are a concise way to create lists.

```python
#traditional loop approach
numbers = [1, 2, 3, 5]
squares_loop = []
for n in numbers:
    squares_loop.append(n ** 2)
print(f"Loop result: {squares_loop}")

#List comprehensiom - same result , one line
squares_comp = [n ** 2 for n in numbers]
print(f"comprehension: {squares_comp}")
```

```python
#List comprehension - same result, one line
squares_comp = [n ** 2 for n in numbers]
print(f"Comprehension: {squares_comp}")
```

```python
#With conditins - filter while transforming
sales = [1200, 15000, 950, 1570, 6900]

big_sales_doubled = [s * 2 for s in sales if s > 100]
print(f"Big sales doubled: {big_sales_doubled}")
```

```python
#Real-Word examples: Extract values from list of dicts
products = [
    {"name": "Laptop", "price": 999}, # Corrected 'names' to 'name'
    {"name": "Mouse", "price": 29},
    {"name": "Keyboard", "price": 89}
]

# Extract all prices
prices = [p["price"] for p in products]
print(f"Prices: {prices}")
print(f"Total: ${sum(prices)}")

# Get names of products over $50
expensive = [p["name"] for p in products if p["price"] > 50]
print(f"Over $50: {expensive}")
```

```python
# Sorting with key function
products = [
    {"name": "Laptop", "price": 999},
    {"name": "Mouse", "price": 29},
    {"name": "Keyboard", "price": 89},

]

#sort by price
by_price = sorted(products, key=lambda p: p["price"])
print("By price (asc):")
for p in by_price:
    print(f"   {p['name']}: ${p['price']}")

#sort by price descending
by_price_desc = sorted(products, key=lambda p: p["price"], reverse=True)
print("\nBy price (desc):")
for p in by_price_desc:
    print(f"   {p['name']}: $ {p['price']}")
```

PART 2: DICTIONARIES DEEP DIVE

Dictionaries are your go-to structure

```python
#create dictionary from two lists
products = ["Laptop", "Mouse", "keyboard"]
prices = [999, 30, 79]

price_dict = {product: price for product, price in zip (products, prices)}

#transform values.
discounted = {k: v * 0.9 for k , v in price_dict.items()}
print(f"10% off: {discounted}")
```

```python
#Filter dictionary
inventory = {"Laptop": 45, "Mouse": 200, "Keyboard": 12, "Monitor": 8}

low_stock = {item: qty for item, qty in inventory.items() if qty < 20}
print(f"Low stock items: {low_stock}")
```

AGGREGATION DATA WITH DICTIONARIES

```python
#Count occurences - a very common pattern
orders = ["Electronic", "Clothing", "Electronics", "Food",
          "Clothing", "Electronics", "Food", "Clothing"]

#Manual Counting
category_counts ={}
for category in orders:
  if category in category_counts:
    category_counts[category] +=1
  else:
      category_counts[category] = 1

  print(f"category counts: {category_counts}")

```

```python
#Group items by a key
transactions = [
    {"date": "2026-01-01", "amount": 100},
    {"date": "2026-01-01", "amount": 250},
    {"date": "2026-01-03", "amount": 175},
    {"date": "2026-01-03", "amount": 80},
    {"date": "2026-01-05", "amount": 400},
]

# Grou[ by date
by_date = {}
for txn in transactions:
  date = txn["date"]
  if date not in by_date.keys():
    by_date[date] = []
  by_date[date].append(txn["amount"])

print("transactions by date:")
for date, amounts in by_date.items():
  print(f"  {date}: {amounts} (total: ${sum(amounts)})")
```
