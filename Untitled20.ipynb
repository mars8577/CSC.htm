LIST COMPREHENSIONS
List comprehension are a concise way to create lists.

```python
#traditional loop approach
numbers = [1, 2, 3, 5]
squares_loop = []
for n in numbers:
    squares_loop.append(n ** 2)
print(f"Loop result: {squares_loop}")

#List comprehensiom - same result , one line
squares_comp = [n ** 2 for n in numbers]
print(f"comprehension: {squares_comp}")
```

```python
#List comprehension - same result, one line
squares_comp = [n ** 2 for n in numbers]
print(f"Comprehension: {squares_comp}")
```

```python
#create dictionary from two lists
products = ["Laptop", "Mouse", "keyboard"]
prices = [999, 30, 79]

price_dict = {product: price for product, price in zip (products, prices)}

#transform values.
discounted = {k: v * 0.9 for k , v in price_dict.items()}
print(f"10% off: {discounted}")
```

```python
#Filter dictionary
inventory = {"Laptop": 45, "Mouse": 200, "Keyboard": 12, "Monitor": 8}

low_stock = {item: qty for item, qty in inventory.items() if qty < 20}
print(f"Low stock items: {low_stock}")
```

```python
#Group items by a key
transactions = [
    {"date": "2026-01-01", "amount": 100},
    {"date": "2026-01-01", "amount": 250},
    {"date": "2026-01-03", "amount": 175},
    {"date": "2026-01-03", "amount": 80},
    {"date": "2026-01-05", "amount": 400},
]

# Grou[ by date
by_date = {}
for txn in transactions:
  date = txn["date"]
  if date not in by_date.keys():
    by_date[date] = []
  by_date[date].append(txn["amount"])

print("transactions by date:")
for date, amounts in by_date.items():
  print(f"  {date}: {amounts} (total: ${sum(amounts)})")
```

Part 3: Nested Data structures
Real World data is rarely flat, it's usually nested - Dictionaries countaining lists dictionairies, and deeper combinations

Common Patterns

```python
# Pattern1: List of dictionaries (like database rows)
customers = [
    {"id": 1, "name": "Alice", "email": "alice@email.com"},
    {"id": 2, "name": "Bob", "email": "bob@email.com"},
    {"id": 3, "name": "Charlie", "email": "charlie@email.com"}
]


 # Access specific customer
print(f"First customer : {customers[0]['name']}")


 # Loop through all
for customer in customers:
  print(f"  {customer['id']}:  {customer['name']}")
```

```python
# Pattern 2: Dictionary of lists
sales_by_region = {
    "North": [12000, 13567, 3400],
    "South": [3459, 7007, 4567],
    "East": [30000, 4575, 90000]
}

print(f"North sales:  {sales_by_region ['North']}")
for region, sales in sales_by_region.items():
  print(f"  {region}: ${sum(sales):,}")
```

```python
#Pattern 3: Deeply nested (common in APIs and JSON)

company = {
    "name": "TechCorps",
    "Departments": [
        {
            "name": "Engineering",
            "employees":[
               {"name": "Alice", "role": "Developer"},
               {"name": "Bob", "role": "Developer"}
            ]
        },
        {
            "name": "Sales",
            "employees": [
                {"name": "Carol", "role": "Manager"},
                {"name": "Dave", "role": "Rep"}
            ]
        }
    ]
}

#Navigate the structure
print(f"Company: {company['name']}")
print(f"First dept: {company['Departments'][0]['name']}")
print(f"First employee in first dept: {company['Departments'][0]['employees'][0]['name']}")

#loop through all employees
print("\nAll employees:")
for dept in company["Departments"]:
    print(f" {dept['name']}")
    for emp in dept['employees']:
        print(f"  - {emp['name']} ({emp['role']})")
```

```python
# Process CSV manualy
employees = []

with open("employees.csv", "r") as f:
   lines = f.readlines()

# The actual header is the second line (index 1) after the initial blank line
header = lines[1].strip().split(",")

# Process data lines, starting from the third line (index 2)
for line in lines[2:]:
    if line.strip(): # Ensure the line is not empty (handles the trailing blank line)
      values = line.strip().split(",")
      employee = {
          "name": values[0].strip(),
          "department": values[1].strip(),
          "salary": int(values[2].strip())
      }
      employees.append(employee)

print(f"Loaded {len(employees)} employees")
for emp in employees:
  print(f" {emp['name']}: ${emp['salary']:,}")
```
